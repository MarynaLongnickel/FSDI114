```python
# Custom Stack Class

class Node:
    def __init__(self, data):
        self.data = data
        self.below = None


class CustomStack:
    def __init__(self):
        self.top = None

    def push(self, value):
        if not self.top:
            self.top = Node(value)
        else:
            new_node = Node(value)
            new_node.below = self.top
            self.top = new_node

    def pop(self):
        if not self.top:
            raise IndexError("Stack is Empty")
        else:
            current = self.top.data
            self.top = self.top.below
            return current

    def is_empty(self):
        return self.top == None

custom_stack = CustomStack()
custom_stack.push(60)
custom_stack.push(70)
custom_stack.push(20)
custom_stack.pop()
custom_stack.pop()
custom_stack.pop()
custom_stack.pop()
```
